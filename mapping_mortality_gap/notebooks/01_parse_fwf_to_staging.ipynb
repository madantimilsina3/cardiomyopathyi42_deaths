```python
#setup & connection

import pandas as pd
import numpy as np
import duckdb
from IPython.display import display

db_path = "/content/drive/MyDrive/mapping_mortality_gap/data_clean/staging.duckdb"
conn = duckdb.connect(db_path)
```

```python
#checking if it works for 2018
check_col_data = [(101,105), (64,66), (68,69), (69,73), (444, 446), (145, 149)]
check_col_name = ["year", "month", "sex", "age", "race", "cause"]
check_file_path = "/content/drive/MyDrive/mapping_mortality_gap/raw_data/microdata/2018/mort2018us.txt"
check_df = pd.read_fwf(check_file_path, colspecs=check_col_data, names=check_col_name, dtype=str)

print(check_df.info())
```

```python
#executing from 2018 to 2024 and connecting to database as table

col_names = ["year", "month", "sex", "age", "race", "cause"]
for year in range(2018, 2025):
  #since tap location for race was changed from 2021
  if year < 2021:
    col_data = [(101,105), (64,66), (68,69), (69,73), (444, 446), (145, 149)]

  else:
    col_data = [(101,105), (64,66), (68,69), (69,73), (488, 490), (145, 149)]

  file_path = f"/content/drive/MyDrive/mapping_mortality_gap/raw_data/microdata/{year}/mort{year}us.txt"
  df = pd.read_fwf(file_path, colspecs=col_data, names = col_names, dtype=str)

  #connecting to database
  conn.execute(f"DROP TABLE IF EXISTS mortality_{year}")
  conn.execute(f"CREATE TABLE mortality_{year} AS SELECT * FROM df")

conn.close()
print(f"All tables added in database successfully")
```
