```python
#connection
import pandas as pd
import numpy as np
import duckdb
from IPython.display import display


db_path = "/content/drive/MyDrive/mapping_mortality_gap/data_clean/staging.duckdb"
conn = duckdb.connect(db_path)
```

```python
# checking no. of rows and unique values in each tables from 2018 to 2024
for i in range(2018, 2025):
  print(f"Details for {i}: ")
  count = conn.execute(f"SELECT COUNT (*) FROM mortality_{i}").fetchone()[0]
  print(f"Number of rows: {count}")
  print(f"Sex values: {conn.execute(f"SELECT DISTINCT sex FROM mortality_{i}").df()["sex"].to_list()}")
  print(f"Months values: {conn.execute(f"SELECT DISTINCT month FROM mortality_{i} ORDER BY month").df()["month"].to_list()}")
  print()

conn.close()

# The output count was almost matched with Official CDC National Death Count with just difference of around 7k to 10k
# Only M and F is perfect
# Month from 1 to 12 is perfect representation of January to December

```
